```python
import pandas as pd

data = {
    'CustomerID': [101, 102, 103, 104, 105, 106, 107, 106],
    'Age': [25, 32, None, 45, 29, 32, 120, 32],  
    'City': ['Dhaka', 'Chittagong', 'Dhaka', 'Dhaka', 'Sylhet', 'Dhaka', 'Dhaka', 'Dhaka'],
    'Purchase_Amount': [500, 1200, 800, None, 1500, 1200, 900, None], 
    'Product': ['Bread', 'Milk', 'Bread', 'Jam', 'Bread', 'Milk', 'Bread', None]
}

df = pd.DataFrame(data)
print(df)
```
