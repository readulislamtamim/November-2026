```python
import requests
import pandas as pd
from bs4 import BeautifulSoup

data = []

headers = {
    "User-Agent": "Mozilla/5.0"
}

for page in [1, 3]:

    url = f"https://www.iplt20.com/news?page={page}"

    response = requests.get(url, headers=headers)

    soup = BeautifulSoup(response.text, "html.parser")

    # MODIFY SELECTOR AFTER INSPECTING HTML
    cards = soup.find_all("div", class_="news-card")

    for card in cards:

        news_type = card.find(
            "div",
            class_="type"
        ).text.strip()

        title = card.find("h2").text.strip()

        date = card.find(
            "div",
            class_="date"
        ).text.strip()

        image = card.find("img")["src"]

        row = {
            "Type": news_type,
            "Title": title,
            "Date": date,
            "Image": image
        }

        data.append(row)

df = pd.DataFrame(data)

print(df.head())

df.to_csv(
    "ipl_news.csv",
    index=False
)
```
