```python
import requests
import pandas as pd
from bs4 import BeautifulSoup
```

### Downloading webpage ###

```python
url = "http://books.toscrape.com/catalogue/page-1.html"
headers = {"User-Agent" : "Mozilla/5.0 (Learning Project)"}

response = requests.get(url, headers = headers, timeout = 10)
response.encoding = 'utf-8' 
print(f"Status: {response.status_code}")
```

### Turning messy data into clear format ###

```python
soup = BeautifulSoup(response.text, "html.parser")
soup
```

### Selecting data ###

```python
soup.find("h1")
```

```python
soup.find_all("h3")
```

```python
soup.find_all("p")
```

### Finding all books ###

```python
books = soup.find_all("article", class_="product_pod")
```

### Collecting each book details ###

```python
data = []
for book in books:
    title = book.h3.a["title"]
    price_text = book.find('p', class_ = "price_color").text
    price = float(price_text.replace('Â', '').replace('£', ''))
    rating = book.find("p", class_ = "star-rating")["class"][1]

    data.append({"Title": title,
                 "Price": price,
                 "Rating" : rating
                })
```

```python
df = pd.DataFrame(data)
df.sample(5)
```

```python
df.describe()
```

### Saving data into csv file ###

```python
df.to_csv("../Datasets/books_scraped.csv")
print(f"Savled {len(df)} books")
```
