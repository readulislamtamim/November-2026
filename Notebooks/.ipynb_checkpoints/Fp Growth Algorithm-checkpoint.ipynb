```python
import pandas as pd

from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import fpgrowth, association_rules
```

### Data loading ###

```python
df = pd.read_csv("../datasets/groceries.csv")
df.head()
```

### Inspecting the data ###

```python
df.shape
```

```python
df.columns.to_list()
```

```python
df.info()
```

### Data cleaning ###

```python
for column in df.columns:
    df[column] = df[column].fillna("")
    df[column] = df[column].astype(str)
    df[column] = df[column].str.strip()
    df[column] = df[column].str.lower()

df
```

### Converting the row data into transaction ###

We will make something like below
<!-- transactions = [
    ['Bread', 'Milk'],
    ['Bread', 'Butter', 'Milk'],
    ['Bread', 'Butter'],
    ['Milk', 'Butter'],
    ['Bread', 'Milk', 'Butter', 'Jam'],
    ['Bread', 'Milk', 'Butter'],
    ['Milk', 'Butter', 'Jam'],
    ['Bread', 'Butter']
] -->


```python
transactions = []

for index, row in df.iterrows():
    basket = []
    for item in row:
        if item != "":
            if item not in basket:
                basket.append(item)

    if len(basket)> 0:
        transactions.append(basket)

#transactions
```

### Converting transaction into boolean table ###

```python
te = TransactionEncoder()
te_array = te.fit(transactions).transform(transactions)

basket_df = pd.DataFrame(
    te_array,
    columns=te.columns_
)
basket_df.sample(5)
```

### Apply PF-Growth ###

```python
frequent_itemsets = fpgrowth(
    basket_df,
    min_support= 0.01,
    use_colnames=True
)

print(frequent_itemsets)
```

```python
# Sorting by support
frequent_itemsets = frequent_itemsets.sort_values(
    by="support",
    ascending=False
)

print("Frequent itemsets:")
print(frequent_itemsets.head(20).to_string(index=False))
```

### Generate association rule ###

```python
rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.30
)

rules = rules[
    [
        "antecedents",
        "consequents",
        "support",
        "confidence",
        "lift"
    ]
]

# Convert frozensets to readable text
rules["antecedents"] = rules["antecedents"].apply(
lambda items: ", ".join(sorted(items))
)

rules["consequents"] = rules["consequents"].apply(
lambda items: ", ".join(sorted(items))
)


# Sort the rules by lift
rules = rules.sort_values(
by="lift",
ascending=False
)

print("\nAssociation rules:")
print(rules.head(20).to_string(index=False))
```

### Explanations ###
Support: 0.010, about 1% of shoping transaction contains all three prodcuts - citrus fruit, other vegetables and root vegetables \
Confidence: Who bought citrus fruit and other vegetables, about 35.9% of them also bought root vegetables. \
Lift: 3.29 which is greater than 1. A strong relationship.
