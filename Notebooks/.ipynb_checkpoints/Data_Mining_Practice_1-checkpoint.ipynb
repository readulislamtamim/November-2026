```python
Q1 = df["Age"].quantile(0.25)
Q3 = df["Age"].quantile(0.75)
IQR = Q3-Q1
lower_bound = Q1 - 1.5*IQR
upper_bound = Q3 + 1.5*IQR

print(f"Q1: {Q1}, Q3: {Q3}, IQR: {IQR}")
print(f"lower_bound: {lower_bound}, upper_bound: {upper_bound}")
```

```python
outlier = []
normal = []
for age in df["Age"]:
    if age < lower_bound or age> upper_bound:
        outlier.append(age)
    else:
        normal.append(age)

print(outlier)
print(normal)
```

```python
df_clean = df[(df["Age"] >= lower_bound) & (df["Age"] <= upper_bound)]
df_clean
```

```python
df["Age"] = df["Age"].clip(lower_bound, upper_bound)
df
```

```python
import pandas as pd

transactions = [
    ['Bread', 'Butter', 'Jam'],
    ['Bread', 'Milk'],
    ['Milk', 'Bread', 'Butter'],
    ['Bread', 'Butter'],
    ['Milk', 'Jam']
]

total_transactions = len(transactions)

# --- REWRITTEN SUPPORT FUNCTION ---
def calculate_support(item_list):
    count = 0
    # Loop through every single shopping basket
    for basket in transactions:
        # Check if every item in our list is inside this basket
        # Example: Is ['Bread', 'Butter'] inside ['Bread', 'Butter', 'Jam']?
        is_present = True
        for item in item_list:
            if item not in basket:
                is_present = False
                break
        
        if is_present:
            count = count + 1
            
    return count / total_transactions

# --- REWRITTEN CONFIDENCE FUNCTION ---
def calculate_confidence(if_this, then_that):
    # 1. Find how many had BOTH (The 'Both' part)
    combined_list = if_this + then_that
    both_count = calculate_support(combined_list)
    
    # 2. Find how many had the first item (The 'If' part)
    if_count = calculate_support(if_this)
    
    # Divide Both by the 'If' part
    if if_count == 0: 
        return 0
    return both_count / if_count

# --- REWRITTEN LIFT FUNCTION ---
def calculate_lift(if_this, then_that):
    # 1. Get Confidence
    conf = calculate_confidence(if_this, then_that)
    
    # 2. Get Support of the second item (The 'Then' part)
    support_that = calculate_support(then_that)
    
    if support_that == 0:
        return 0
    return conf / support_that

# --- TESTING THE CODE ---

# Rule: {Bread} -> {Butter}
print("--- Rule: Bread -> Butter ---")
print(f"Support: {calculate_support(['Bread', 'Butter']):.2f}")
print(f"Confidence: {calculate_confidence(['Bread'], ['Butter']):.2f}")
print(f"Lift: {calculate_lift(['Bread'], ['Butter']):.2f}")

# Rule: {Bread} -> {Jam}
print("\n--- Rule: Bread -> Jam ---")
print(f"Support: {calculate_support(['Bread', 'Jam']):.2f}")
print(f"Confidence: {calculate_confidence(['Bread'], ['Jam']):.2f}")
print(f"Lift: {calculate_lift(['Bread'], ['Jam']):.2f}")
```

## Lesson 1: Market Basket Analysis + Support, Confidence, Lift ##

```python
import pandas as pd

transactions = [
    ['Bread', 'Butter', 'Jam'],
    ['Bread', 'Milk'],
    ['Milk', 'Bread', 'Butter'],
    ['Bread', 'Butter'],
    ['Milk', 'Jam']
]

total_transactions = len(transactions)
print(total_transactions)

# Support function
def calculate_support(item_list):
    count = 0
    for basket in transactions:
        is_present = True
        for item in item_list:
            if item not in basket:
                is_present = False
                break
        if is_present:
            count = count + 1
    return count / total_transactions

# Confidence function
def calculate_confidence(if_that, then_that):
    combined_list = if_that + then_that
    both_count = calculate_support(combined_list)

    if_count = calculate_support(if_that)

    if if_count == 0:
        return 0
    else:
        return both_count / if_count

def calculate_lift(if_that, then_that):
    conf = calculate_confidence(if_that, then_that)
    support_that = calculate_support(then_that)

    if support_that == 0:
        return 0
    else:
        return conf/support_that
    

calculate_support_result = calculate_support(['Bread', 'Butter'])
calculate_confidence_result = calculate_confidence(["Bread"], ["Butter"])


print("--- Rule: Bread -> Butter ---")
print(f"Support: {calculate_support(['Bread', 'Butter']):.2f}")
print(f"Confidence: {calculate_confidence(['Bread'], ['Butter']):.2f}")
print(f"Lift: {calculate_lift(['Bread'], ['Butter']):.2f}")

print("\n--- Rule: Bread -> Jam ---")
print(f"Support: {calculate_support(['Bread', 'Jam']):.2f}")
print(f"Confidence: {calculate_confidence(['Bread'], ['Jam']):.2f}")
print(f"Lift: {calculate_lift(['Bread'], ['Jam']):.2f}")
```
