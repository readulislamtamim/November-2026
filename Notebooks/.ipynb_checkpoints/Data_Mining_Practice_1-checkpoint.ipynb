```python
ages = [25, 35, 40, 18, 27, 39, 120]
df = pd.DataFrame({"Age": ages})

df
```

```python
Q1 = df["Age"].quantile(0.25)
Q3 = df["Age"].quantile(0.75)
IQR = Q3-Q1
lower_bound = Q1 - 1.5*IQR
upper_bound = Q3 + 1.5*IQR

print(f"Q1: {Q1}, Q3: {Q3}, IQR: {IQR}")
print(f"lower_bound: {lower_bound}, upper_bound: {upper_bound}")
```

```python
outlier = []
normal = []
for age in df["Age"]:
    if age < lower_bound or age> upper_bound:
        outlier.append(age)
    else:
        normal.append(age)

print(outlier)
print(normal)
```

```python
df_clean = df[(df["Age"] >= lower_bound) & (df["Age"] <= upper_bound)]
df_clean
```

```python
df["Age"] = df["Age"].clip(lower_bound, upper_bound)
df
```

```python
df = pd.DataFrame({"Age" : [25, 32, 32, 45, 29, 32, 120]})
df
```

```python
#Z = (x-mean)/std
df["Z_Score"] = (df["Age"] - df["Age"].mean())/df["Age"].std()
df
```

```python
outliers_z = df[abs(df["Z_Score"]) > 3]
print(f"Z_Score outliers: {outliers_z}")
```

## 1. Concatenation - Stacking (Vertical or Horizontal) ##

```python
import pandas as pd
```

```python
df_jan = pd.DataFrame({
    "Customer_ID": [101, 102],
    "Purchase_Amount": [500, 1300]
})

df_jan
```

```python
df_feb = pd.DataFrame({
    "Customer_ID": [103, 104],
    "Purchase_Amount": [800, 1000]
})
df_feb
```

```python
df_all = pd.concat([df_jan, df_feb], ignore_index=True)
df_all
```

## 2. Merging / Joining - The Most Important ##

```python
import pandas as pd
```

```python
customers = pd.DataFrame({
    'CustomerID': [101, 102, 103, 104],
    'Name': ['Amar', 'Rahim', 'Karim', 'Nusrat'],
    'City': ['Dhaka', 'Chittagong', 'Dhaka', 'Sylhet']
})
customers
```

```python
orders = pd.DataFrame({
    'OrderID': [1, 2, 3, 4, 5],
    'CustomerID': [101, 101, 102, 103, 105], # Note: 105 is not in customers, 104 has no order
    'Product': ['Bread', 'Milk', 'Jam', 'Bread', 'Bread'],
    'Amount': [500, 800, 1200, 900, 400]
})
orders
```

```python
inner_join = pd.merge(customers, orders, on = "CustomerID", how = "inner")
inner_join
```

```python
left_join = pd.merge(customers, orders, on = "CustomerID", how = "left")
left_join
```

```python
right_join = pd.merge(customers, orders, on = "CustomerID", how = "right")
right_join
```

```python
outer_join = pd.merge(customers, orders, on = "CustomerID", how = "outer")
outer_join
```

## 1. Normalization - Min-Max Scaling (0 to 1) ##

```python
# (x-min)/(max-min)
```

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
```

```python
df = pd.DataFrame({
    'Age': [25, 32, 45, 29],
    'Purchase_Amount': [500, 1200, 1500, 800]
})

df
```

```python
scaler = MinMaxScaler()
df_scaled = pd.DataFrame(scaler.fit_transform(df), columns=df.columns)
df_scaled
```

## 2. Standardization - Z-Score Scaling (Mean=0, Std=1) ##

```python
# (x - mean) / std
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.DataFrame({
    'Age': [25, 32, 45, 29],
    'Purchase_Amount': [500, 1200, 1500, 800]
})

df
```

```python
scaler = StandardScaler()
df_std = pd.DataFrame(scaler.fit_transform(df), columns=df.columns)
df_std
```
